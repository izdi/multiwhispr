# Spectrograms of the last two MultiWhispr recordings

Left channel = you, right channel = other apps.

```python
import json
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
import warnings; warnings.filterwarnings("ignore")

meetings = Path.home() / "Library/Application Support/MultiWhispr/Meetings"
recordings = sorted(p for p in meetings.iterdir() if (p / "recording.wav").exists())[-2:]
[r.name for r in recordings]
```

```python
def show(folder):
    rate, data = wavfile.read(folder / "recording.wav")
    meta = json.loads((folder / "session.json").read_text())
    t = np.arange(len(data)) / rate
    fig, axes = plt.subplots(2, 1, figsize=(14, 7), sharex=True)
    for ax, ch, name in zip(axes, (0, 1), ("left = you", "right = others")):
        x = data[:, ch] / 32768
        ax.specgram(x, Fs=rate, NFFT=512, noverlap=384, cmap="magma", vmin=-120, vmax=-20)
        peak = float(np.abs(x).max())
        ax.set_title(f"{folder.name} — {name} (peak {peak:.3f}, duration {meta['duration']:.1f}s)")
        ax.set_ylabel("Hz")
    axes[1].set_xlabel("seconds")
    plt.tight_layout()
    plt.show()

for folder in recordings:
    show(folder)
```
